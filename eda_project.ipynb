```python
df['PoolQC'] = df['PoolQC'].fillna('None')
```

```python
df['Alley'] = df['Alley'].fillna('None')
df['Fence'] = df['Fence'].fillna('None')
df['FireplaceQu'] = df['FireplaceQu'].fillna('None')
df['MiscFeature'] = df['MiscFeature'].fillna('None')
df['MasVnrType'] = df['MasVnrType'].fillna('None')
df['GarageType'] = df['GarageType'].fillna('None')
df['GarageFinish'] = df['GarageFinish'].fillna('None')
df['GarageQual'] = df['GarageQual'].fillna('None')
df['GarageCond'] = df['GarageCond'].fillna('None')
df['BsmtQual'] = df['BsmtQual'].fillna('None')
df['BsmtCond'] = df['BsmtCond'].fillna('None')
df['BsmtExposure'] = df['BsmtExposure'].fillna('None')
df['BsmtFinType1'] = df['BsmtFinType1'].fillna('None')
df['BsmtFinType2'] = df['BsmtFinType2'].fillna('None')
```

```python
df['LotFrontage'] = df['LotFrontage'].fillna(df['LotFrontage'].median())
df['MasVnrArea'] = df['MasVnrArea'].fillna(df['MasVnrArea'].median())
df['GarageYrBlt'] = df['GarageYrBlt'].fillna(df['GarageYrBlt'].median())
df['Electrical'] = df['Electrical'].fillna(df['Electrical'].mode()[0])
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
sns.histplot(df['SalePrice'], kde=True)
plt.title('Distribution of SalePrice')
plt.show()
```

```python
import numpy as np
```

 Skewness before the transform


```python
print("Skew before:", df['SalePrice'].skew())
```

 log1p = log(1 + x), the standard safe way to log-transform prices


```python
df['SalePrice_log'] = np.log1p(df['SalePrice'])
```

 Skewness after


```python
print("Skew after:", df['SalePrice_log'].skew())
```

```python
sns.histplot(df['SalePrice_log'], kde=True)
plt.title('Distribution of log(SalePrice)')
plt.show()
```

Next: which features relate to SalePrice?

 Correlation of every numeric column with SalePrice, strongest first


```python
corr = df.select_dtypes(include='number').corr()['SalePrice'].sort_values(ascending=False)
corr.head(11)
```

 Scatter plot: living area vs price


```python
sns.scatterplot(x=df['GrLivArea'], y=df['SalePrice'])
plt.title('GrLivArea vs SalePrice')
plt.show()
```

 Boxplot: overall quality rating vs price


```python
sns.boxplot(x=df['OverallQual'], y=df['SalePrice'])
plt.title('OverallQual vs SalePrice')
plt.show()
```

# Very large houses that sold cheapl


```python
df[(df['GrLivArea'] > 4000) & (df['SalePrice'] < 300000)]
```

 Drop the 2 outlier rows (large area but very low price)


```python
df = df.drop(df[(df['GrLivArea'] > 4000) & (df['SalePrice'] < 300000)].index)
```

```python
print(df.shape)
```

```python
sns.scatterplot(x=df['GrLivArea'], y=df['SalePrice'])
plt.title('GrLivArea vs SalePrice (after removing outliers)')
plt.show()
```

```python
top_cols = ['SalePrice', 'OverallQual', 'GrLivArea', 'GarageCars', 'GarageArea', 'TotalBsmtSF', '1stFlrSF', 'YearBuilt']
sns.heatmap(df[top_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation of Top Features')
plt.show()
```

```python
# Median price per neighborhood, highest first
df.groupby('Neighborhood')['SalePrice'].median().sort_values(ascending=False)
```

```python
plt.figure(figsize=(12, 5))
sns.boxplot(x=df['Neighborhood'], y=df['SalePrice'])
plt.xticks(rotation=90)
plt.title('SalePrice by Neighborhood')
plt.show()
```
